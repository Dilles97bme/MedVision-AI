## Dataset Path

Instead of hardcoding absolute paths throughout the notebook, we define a single dataset path using pathlib.

This makes the notebook portable and easier to maintain.

```python
from pathlib import Path

DATASET_PATH = Path(r"C:\MedVision-AI\dataset\chest_xray")
```

```python
for folder in DATASET_PATH.iterdir():
    if folder.is_dir():
        print(folder.name)
```

```python
for split in DATASET_PATH.iterdir():

    if split.is_dir():

        print(f"\n{split.name}")

        for cls in split.iterdir():

            if cls.is_dir():

                print("   ", cls.name)
```

## Interpretation

The dataset follows a standard image-classification directory structure.

Each dataset split contains two classes:

- NORMAL
- PNEUMONIA

The dataset already includes train, validation, and test folders.

However, the validation set is extremely small (16 images), so we will later create a new validation split from the training data.
